```python
import pandas as pd
import sqlite3

conn = sqlite3.connect('football.db')

files = ['appearances', 'games', 'leagues', 'players', 'shots', 'teams', 'teamstats']

for file in files:
    df = pd.read_csv(f'{file}.csv', encoding='latin-1')
    df.to_sql(file, conn, if_exists='replace', index=False)
    print(f'{file} loaded ✓')

conn.close()
print('Done! football.db created')
```

```python
%%sql
SELECT t.name, 
       COUNT(*) as total_games,
       SUM(ts.wins) as total_wins
FROM teamstats ts
JOIN teams t ON ts.teamID = t.teamID
GROUP BY t.name
ORDER BY total_wins DESC
LIMIT 10;
```

```python
%%sql
SELECT t.name,
       COUNT(*) as total_games,
       SUM(CASE WHEN ts.result = 'W' THEN 1 ELSE 0 END) as total_wins
FROM teamstats ts
JOIN teams t ON ts.teamID = t.teamID
GROUP BY t.name
ORDER BY total_wins DESC
LIMIT 10;
```

```python
%%sql
SELECT t.name,
       SUM(ts.goals) as total_goals
FROM teamstats ts
JOIN teams t ON ts.teamID = t.teamID
GROUP BY t.name
ORDER BY total_goals DESC
LIMIT 10;
```
